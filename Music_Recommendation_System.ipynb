Text preprocessing or Text Cleaing

```python
df['text']=df['text'].str.lower()
```

```python
# Only lowercasing and removing newlines, keep words for vectorization
```

```python
df['text'] = df['text'].replace(r'\n', ' ', regex=True)
```

```python
df.head(10)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

nltk.download('punkt')
```

```python
stemmer=PorterStemmer()
```

```python
def token(txt):
  token=nltk.word_tokenize(txt)
  a=[stemmer.stem(w) for w in token]
  return " ".join(a)
```

```python
token("you are beautiful, beauty")
```

```python
# Apply stemming and tokenization, and save back to df['text']
df['text'] = df['text'].apply(lambda x: token(x))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity # Correct the module name here
```

```python
tfid=TfidfVectorizer(analyzer='word',stop_words='english')
```

```python
matrix = tfid.fit_transform(df['text'])
```

```python
print('Sample of df["text"] after preprocessing and stemming:')
print(df['text'].head(20))
```

```python
# Filter out empty or null text rows before vectorization
df = df[df['text'].notnull() & (df['text'].str.strip() != '')]
```

```python
similar=cosine_similarity(matrix)
```

```python
similar[0]
```

Recommender Function

```python
def recommender(song_name, df):
    if 'song' not in df.columns:
        print("Error: 'song' column not found in DataFrame.")
        return []
    
    if song_name not in df['song'].values:
        print(f"Song '{song_name}' not found in DataFrame.")
        return []
    
    idx = df[df['song'] == song_name].index[0]
    
    distance = sorted(list(enumerate(similar[idx])), reverse=True, key=lambda x: x[1])
    
    song = []
    
    for s_id in distance[1:21]:
        song.append(df.iloc[s_id[0]]['song'])
    
    return song

# Example usage:
print(recommender("Can't You See", df))
```

```python
# Try the recommender with another song name
print(recommender('God Gave My Everything', df))
```
